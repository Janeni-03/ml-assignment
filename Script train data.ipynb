```python
import os, shutil, zipfile

os.makedirs("breast_cancer_data", exist_ok=True)

# if you uploaded wdbc.data directly
if os.path.exists("wdbc.data") and not os.path.exists("breast_cancer_data/wdbc.data"):
    shutil.move("wdbc.data", "breast_cancer_data/wdbc.data")

# if you uploaded the dataset zip
for z in [f for f in os.listdir() if f.endswith(".zip") and f != "model_files.zip"]:
    with zipfile.ZipFile(z) as zf:
        zf.extractall("breast_cancer_data")

print(os.listdir("breast_cancer_data"))
```
